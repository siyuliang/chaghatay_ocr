```python
from jiwer import wer
import editdistance

def calculate_cer(gold_text, ocr_text):
    """
    Calculate the Character Error Rate (CER).
    """
    gold_chars = list(gold_text.replace("\n", ""))
    ocr_chars = list(ocr_text.replace("\n", ""))
    
    # Calculate edit distance
    cer = editdistance.eval(gold_chars, ocr_chars) / len(gold_chars)
    return cer

# Load the gold standard and OCR output
with open('../data/test/gold.txt', 'r', encoding='utf-8') as f:
    gold_text = f.read()

with open('../data/test/newoutput.txt', 'r', encoding='utf-8') as f:
    ocr_text = f.read()

# Calculate WER and CER
word_error_rate = wer(gold_text, ocr_text)
character_error_rate = calculate_cer(gold_text, ocr_text)

print(f"Word Error Rate (WER): {word_error_rate:.2%}")
print(f"Character Error Rate (CER): {character_error_rate:.2%}")
```
